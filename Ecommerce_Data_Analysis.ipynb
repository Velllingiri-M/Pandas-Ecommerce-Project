```python
data.rename(
    columns={"Description": "ProductDescription"},
    inplace=True
)

print(data.columns)
```

```python
data["Revenue"] = data["Quantity"] * data["UnitPrice"]

print(data[["Quantity", "UnitPrice", "Revenue"]].head())
```

```python
import numpy as np
conditions = [
    data["InvoiceNo"].str.startswith("C", na=False),
    data["Quantity"] < 0,
    (data["Quantity"] > 0) & (data["UnitPrice"] > 0)
]

choices = ["Cancellation", "Return", "Positive sale"]

data["TransactionType"] = np.select(
    conditions,
    choices,
    default="Other/invalid"
)

print(data[["InvoiceNo", "Quantity", "UnitPrice", "TransactionType"]])
```

```python
data["IsIdentifiedCustomer"] = data["CustomerID"].notna()

print(data[["CustomerID", "IsIdentifiedCustomer"]])
```

```python
data["InvoiceDate"] = pd.to_datetime(
    data["InvoiceDate"],
    errors="coerce"
)

print(data["InvoiceDate"])
print(data["InvoiceDate"].dtype)
```

```python
data["InvoiceNo"] = data["InvoiceNo"].astype("string")
data["StockCode"] = data["StockCode"].astype("string")
data["Country"] = data["Country"].astype("string")

print(data[["InvoiceNo", "StockCode", "Country"]].dtypes)
```

```python
data = data[[
    "InvoiceNo",
    "InvoiceDate",
    "StockCode",
    "ProductDescription",
    "CustomerID",
    "Country",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType",
    "IsIdentifiedCustomer"
]]

print(data.head())
```

```python
data = data.sort_values(
    by=["InvoiceDate", "InvoiceNo"],
    ascending=[True, True]
)

print(data)
```

```python
data = data.reset_index(drop=True)

print(data)
print("Index:", data.index.tolist())
print("Sequential index:", data.index.equals(pd.RangeIndex(len(data))))
```
